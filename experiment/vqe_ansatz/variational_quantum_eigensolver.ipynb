```python
from qiskit.circuit.library import TwoLocal
from qiskit import QuantumCircuit
import matplotlib.pyplot as plt

# Optional: Set LaTeX + Times New Roman style
plt.rcParams.update({
    "text.usetex": False,
    "font.family": "serif",
    "font.serif": ["Liberation Serif", "DejaVu Serif", "Georgia", "Cambria"]
})

import matplotlib.pyplot as plt
from matplotlib import rcParams

# Force all matplotlib fonts to use Times New Roman
rcParams['font.family'] = 'serif'
rcParams['font.serif'] = ['Times New Roman']
rcParams['text.usetex'] = False  # safer for Unicode like θ

# Create the ansatz
vqe_circuit = TwoLocal(num_qubits=3, rotation_blocks='ry', entanglement_blocks='cz',
                       entanglement='full', reps=1, insert_barriers=True)

# Decompose to show individual gates
decomposed_circuit = vqe_circuit.decompose()

# Add measurements to all qubits
decomposed_circuit.measure_all()

# Draw and save
fig = decomposed_circuit.draw(output='mpl', filename='vqe_circuit_with_measure.pdf')
plt.close()
```
