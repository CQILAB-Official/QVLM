````python
import csv
from qiskit import QuantumCircuit

# --- Load CSV file ---
base_path = "../../../../"
goto_folder = "ResultGroup/3.Circuit/v3/"
filename = "Circuit-QuantumVLM-2.5VL-7B-v3-focused-compile.1590.csv"
csv_path = f"{base_path}{goto_folder}{filename}"
column_name = "prediction_5"        # column with QASM strings

success_count = 0
fail_count = 0
codes = []

def clean_prediction(text):
    # Replace ```qasm with OPENQASM 2.0;
    text = text.replace("```qasm", "OPENQASM 2.0;")
    # Remove trailing ```
    if text.endswith("```"):
         text = text[:-3]
    return text.strip()

# --- Load CSV ---
with open(csv_path, "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        codes.append(clean_prediction(row[column_name]))

# --- Test each QASM code ---
results = []
for idx, code in enumerate(codes, start=1):
    try:
        # Parse QASM (works for OpenQASM 2.0)
        qc = QuantumCircuit.from_qasm_str(code)
        results.append((idx, True))
        success_count += 1
    except Exception as e:
        # Any parse or build error → fail
        results.append((idx, False))
        fail_count += 1

# --- Summary ---
total = len(codes)
runnability = success_count / total if total > 0 else 0

print("Total samples:", total)
print("Success:", success_count)
print("Fail:", fail_count)
print("Runnability:", runnability)

for idx, ok in results:
    print(f"Row {idx}: {'OK' if ok else 'ERROR'}")
````
