```python
import csv
import numpy as np
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report

# Function to read data from CSV
def read_csv(file_path):
    output_6 = []
    prediction_6 = []
    
    with open(file_path, mode='r', encoding='utf-8') as file:
        reader = csv.DictReader(file)
        for row in reader:
            output_6.append(row['output_6'])  # Ground truth
            prediction_6.append(row['prediction_6'])  # Predictions
    
    return output_6, prediction_6

# Read ground truth and predictions from the CSV file
base_path = "../../../"
goto_folder = "ResultGroup/3.Circuit/"
filename = "Circuit-QuantumVLM-2.5VL-32B-v2.1060.csv"
csv_path = f"{base_path}{goto_folder}{filename}"
output_6, prediction_6 = read_csv(csv_path)

# Ensure the data is compatible (both lists should have the same length)
if len(output_6) != len(prediction_6):
    raise ValueError("The length of output_6 and prediction_6 must be the same.")

# Get unique classes from the ground truth and prediction
unique_classes = list(set(output_6))  # Union of unique classes

print("Unique Classes:", unique_classes)

# Calculate accuracy
accuracy = accuracy_score(output_6, prediction_6)
print(f"Overall Accuracy: {accuracy:.4f}")

# Calculate precision and recall for each class
precision = precision_score(output_6, prediction_6, labels=unique_classes, average=None, zero_division=0)
recall = recall_score(output_6, prediction_6, labels=unique_classes, average=None, zero_division=0)

# Calculate mean precision and recall
mean_precision = np.mean(precision)
mean_recall = np.mean(recall)

print(f"\nMean Precision: {mean_precision:.4f}")
print(f"Mean Recall: {mean_recall:.4f}")

# Calculate accuracy per class
accuracy_per_class = {}
for class_name in unique_classes:
    true_positives = sum(1 for o, p in zip(output_6, prediction_6) if o == class_name and p == class_name)
    class_total = output_6.count(class_name)
    accuracy_per_class[class_name] = true_positives / class_total if class_total > 0 else 0.0

print("\nAccuracy per Class:")
for class_name, accuracy_value in accuracy_per_class.items():
    print(f"{class_name}: {accuracy_value:.4f}")

# Print Precision and Recall for each class
print("\nPrecision per Class:")
for class_name, precision_value in zip(unique_classes, precision):
    print(f"{class_name}: {precision_value:.4f}")

print("\nRecall per Class:")
for class_name, recall_value in zip(unique_classes, recall):
    print(f"{class_name}: {recall_value:.4f}")

# Print the classification report (including precision, recall, and F1-score)
print("\nClassification Report:")
print(classification_report(output_6, prediction_6, labels=unique_classes))
```
