```python
import csv

# ----- Load CSV -----
base_path = "../"
goto_folder = "2.Output/v2/"
filename = "3.Converted-Wigner-QuantumVLM-2.5VL-7B-v2444-v2.csv"

path = f"{base_path}{goto_folder}{filename}"

with open(path, "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    rows = list(reader)

# ----- Numeric-safe comparison -----
def same_value(a, b):
    """Returns True if a and b represent the same number or same string."""
    a = str(a).strip()
    b = str(b).strip()

    # Try numeric compare
    try:
        return float(a) == float(b)
    except ValueError:
        # fallback to strict string compare
        return a == b


# ----- Counters -----
count = len(rows)
correct_state = 0
correct_param = 0
correct_dimension = 0
correct_qubits = 0
correct_linear = 0
correct_all = 0

for r in rows:
    # Read generated values
    sg = r["state_generated"].strip()
    pg = r["param_generated"].strip()
    dg = r["dimension_generated"].strip()
    qg = r["qubits_generated"].strip()
    lg = r["linear_generated"].strip()

    # Read ground truth values
    sg_t = r["state_gt"].strip()
    pg_t = r["param_gt"].strip()
    dg_t = r["dimension_gt"].strip()
    qg_t = r["qubits_gt"].strip()
    lg_t = r["linear_gt"].strip()

    # Compare
    state_ok = (sg == sg_t)
    param_ok = same_value(pg, pg_t)
    dimension_ok = same_value(dg, dg_t)
    qubits_ok = same_value(qg, qg_t)
    linear_ok = same_value(lg, lg_t)

    # Count each
    correct_state += state_ok
    correct_param += param_ok
    correct_dimension += dimension_ok
    correct_qubits += qubits_ok
    correct_linear += linear_ok

    # Count all-match
    if state_ok and param_ok and dimension_ok and qubits_ok:
        correct_all += 1


# ----- Display results -----
def pct(x):
    return 100 * x / count if count else 0

print("Total:", count)
print(f"State accuracy:  {pct(correct_state):.2f}%")
print(f"Param accuracy:  {pct(correct_param):.2f}%")
print(f"Dimension accuracy: {pct(correct_dimension):.2f}%")
print(f"Qubits accuracy: {pct(correct_qubits):.2f}%")
print(f"Linear accuracy: {pct(correct_linear):.2f}%")
print(f"All correct:     {pct(correct_all):.2f}%")
```
