```python
import os
import pandas as pd

target_dir = './ResultGroup/2.Entanglement'
print(f"Checking directory: {target_dir}")

if not os.path.exists(target_dir):
    print(f"Directory not found: {os.path.abspath(target_dir)}")
    exit(1)

for filename in os.listdir(target_dir):
    full_path = os.path.join(target_dir, filename)
    if os.path.isfile(full_path) and filename.endswith('.csv'):
        try:
            # Use pandas to properly handle quoted newlines
            df = pd.read_csv(full_path)
            print(f"{filename}: {len(df)} rows")
        except Exception as e:
            print(f"Error reading {filename}: {e}")
```
