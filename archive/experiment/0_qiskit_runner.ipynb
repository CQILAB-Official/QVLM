```python
from qiskit_aer import Aer
from qiskit import transpile
from qiskit.qasm2 import loads
from qiskit.visualization import plot_histogram

# QASM program
qasm_code = """
OPENQASM 2.0;
include "qelib1.inc";

qreg q[4];
creg c[4];

// Apply Hadamard gates to qubits q0, q1, and q2
h q[0];
h q[1];
h q[2];

// Apply controlled-Z gates
cz q[0], q[1];
cz q[1], q[2];

// Apply Rz gates with angles -π/4, -π/2, and π/4
rz(-pi/4) q[0];
rz(-pi/2) q[1];
rz(pi/4) q[3];

// Apply another Hadamard gate to qubit q3
h q[3];

// Apply another controlled-Z gate
cz q[2], q[3];

// Measure all qubits into classical bits
measure q[0] -> c[0];
measure q[1] -> c[1];
measure q[2] -> c[2];
measure q[3] -> c[3];
"""

# Load circuit from QASM
qc = loads(qasm_code)

# Choose simulator backend
simulator = Aer.get_backend("qasm_simulator")

# Transpile before running
compiled_circuit = transpile(qc, simulator)

# Run with 1024 shots
job = simulator.run(compiled_circuit, shots=1024)
result = job.result()
counts = result.get_counts()

print("Measurement counts:", counts)

# (Optional) plot histogram
# plot_histogram(counts).show()
```
