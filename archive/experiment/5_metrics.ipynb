```python
from pprint import pprint
import json

from torchmetrics.text.bert import BERTScore
from torchmetrics.text import BLEUScore
from torchmetrics.text import CharErrorRate
from torchmetrics.text import MatchErrorRate
from torchmetrics.text import WordErrorRate
from torchmetrics.multimodal.clip_score import CLIPScore
from torchmetrics.text import Perplexity

# BERT SCORE

# Load from Result/1_cot_result.json column generated and ground_truth each
preds = []
target = []

with open('Result/1_cot_result.json', 'r') as f:
    data = json.load(f)

for i in range(0, len(data)):
    preds.append(data[i]['generated'])
    target.append(data[i]['ground_truth'])

bertscore = BERTScore()
bert_score = bertscore(preds, target)
print("Full BERTScore output:")
pprint(bert_score)

# Calculate and print averages
avg_f1 = bert_score['f1'].mean().item()
avg_precision = bert_score['precision'].mean().item()
avg_recall = bert_score['recall'].mean().item()

print(f"\nAverage F1 Score    : {avg_f1:.4f}")
print(f"Average Precision     : {avg_precision:.4f}")
print(f"Average Recall        : {avg_recall:.4f}")

wer = WordErrorRate()
pprint(wer(preds, target))

# BLEU SCORE 1gram
bleu = BLEUScore(n_gram=1)
pprint(bleu(preds, target))

# BLEU SCORE 2gram
bleu = BLEUScore(n_gram=2)
pprint(bleu(preds, target))

# CER
cer = CharErrorRate()
pprint(cer(preds, target))

# MER
mer = MatchErrorRate()
pprint(mer(preds, target))

# WER
wer = WordErrorRate()
pprint(wer(preds, target))

# print("Original Model")
# preds_org = [" MRI of the brain showing a large lesion in the left frontal lobe.", 
#          "CT scan of the abdomen showing a large mass in the right upper quadrant."]

# bertscore = BERTScore()
# pprint(bertscore(preds_org, target))

# # BLEU SCORE 1gram
# bleu = BLEUScore(n_gram=1)
# pprint(bleu(preds_org, target))

# # BLEU SCORE 2gram
# bleu = BLEUScore(n_gram=2)
# pprint(bleu(preds_org, target))

# # CER
# cer = CharErrorRate()
# pprint(cer(preds_org, target))

# # MER
# mer = MatchErrorRate()
# pprint(mer(preds_org, target))

# # WER
# wer = WordErrorRate()
# pprint(wer(preds_org, target))

# print("BLIP-2")
# preds_blip2 = ["a mri image of a man's head and brain", 
#          "a black and white image of a man's stomach"]

# bertscore = BERTScore()
# pprint(bertscore(preds_blip2, target))

# # BLEU SCORE 1gram
# bleu = BLEUScore(n_gram=1)
# pprint(bleu(preds_blip2, target))

# # BLEU SCORE 2gram
# bleu = BLEUScore(n_gram=2)
# pprint(bleu(preds_blip2, target))

# # CER
# cer = CharErrorRate()
# pprint(cer(preds_blip2, target))

# # MER
# mer = MatchErrorRate()
# pprint(mer(preds_blip2, target))

# # WER
# wer = WordErrorRate()
# pprint(wer(preds_blip2, target))


# CLIP SCORE
# metric = CLIPScore(model_name_or_path="openai/clip-vit-base-patch16")
# values = [ ]
# for _ in range(10):
#    values.append(metric(torch.randint(255, (3, 224, 224)), "a photo of a cat"))
#fig_, ax_ = metric.plot(values)
```
