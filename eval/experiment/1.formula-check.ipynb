```python
# pip install qiskit qiskit-aer pylatexenc matplotlib

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# Your exact OpenQASM circuit
qasm_str = """
OPENQASM 2.0;
include "qelib1.inc";
qreg q[2];
creg c[2];
h q[0];
cx q[0], q[1];
measure q -> c;
"""

# Load and run
qc = QuantumCircuit.from_qasm_str(qasm_str)

# --- ADDED LINE TO GENERATE PDF ---
# This saves the circuit as 'my_circuit.pdf' using the Matplotlib renderer
qc.draw(output='mpl', filename='my_circuit.pdf')
# ----------------------------------

simulator = AerSimulator()
counts = simulator.run(qc, shots=1024).result().get_counts()

print("Counts:", counts)
print("Circuit saved as my_circuit.pdf")
```
