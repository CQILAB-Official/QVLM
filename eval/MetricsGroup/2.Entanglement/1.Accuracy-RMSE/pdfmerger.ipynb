```python
from pypdf import PdfReader, PdfWriter, PageObject

# Input PDFs (order = top → bottom)
pdf_files = [
    "2-1-1.accuracy-entanglement-baseline-1.pdf",
    "2-1-2.accuracy-entanglement-baseline-2.pdf",
    "2-1-3.accuracy-entanglement-quantumvlm-2.5vl.pdf",
    "2-1-4.accuracy-entanglement-quantumvlm-3vl.pdf",
]

writer = PdfWriter()
pages = [PdfReader(f).pages[0] for f in pdf_files]

# Assume all pages have the same size
w = float(pages[0].mediabox.width)
h = float(pages[0].mediabox.height)

# Create a blank page tall enough for all plots
merged_page = PageObject.create_blank_page(
    width=w,
    height=h * len(pages)
)

# Place pages top → bottom
y_offset = h * (len(pages) - 1)
for page in pages:
    merged_page.merge_translated_page(page, 0, y_offset)
    y_offset -= h

writer.add_page(merged_page)

with open("merged_column.pdf", "wb") as f:
    writer.write(f)

print("Saved merged_column.pdf")
```
