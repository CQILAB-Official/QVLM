```python
import csv
from torchmetrics.text.bert import BERTScore

# --- Load CSV file ---
base_path = "../../../../"
goto_folder = "ResultGroup/3.Circuit/"
filename = "Circuit-Baseline2-Llama.csv"

with open(f"{base_path}{goto_folder}{filename}", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    data = list(reader)

pred_cols = [f"prediction_{i}" for i in range(1, 5)]
tgt_cols  = [f"output_{i}"     for i in range(1, 5)]

print("=== Combined BERTScore Evaluation ===")

bertscore = BERTScore(max_length=512, truncation=True)

def sanitize(t):
    t = str(t).strip()
    return t if t else "null"

# ---------------------------------------------------------
# Collect ALL predictions and ALL outputs into one list
# ---------------------------------------------------------
overall_preds = []
overall_tgts  = []

for pcol, tcol in zip(pred_cols, tgt_cols):
    preds = [sanitize(row[pcol]) for row in data]
    tgts  = [sanitize(row[tcol]) for row in data]

    overall_preds.extend(preds)
    overall_tgts.extend(tgts)

print(f"Total combined samples: {len(overall_preds)}")

# ---------------------------------------------------------
# Compute ONE overall BERTScore
# ---------------------------------------------------------
try:
    bert = bertscore(overall_preds, overall_tgts)

    f1 = bert["f1"].mean().item()
    precision = bert["precision"].mean().item()
    recall = bert["recall"].mean().item()

    print("\n=== FINAL COMBINED BERTSCORE ===")
    print("F1:", f1)
    print("Precision:", precision)
    print("Recall:", recall)

except Exception as e:
    print("Combined BERTScore failed:", e)
```

```python
import csv
from torchmetrics.text import BLEUScore

# --- Load CSV file ---
base_path = "../../../../"
goto_folder = "ResultGroup/3.Circuit/"
filename = "Circuit-Baseline2-Llama.csv"

with open(f"{base_path}{goto_folder}{filename}", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    data = list(reader)

pred_cols = [f"prediction_{i}" for i in range(1, 5)]
tgt_cols  = [f"output_{i}"     for i in range(1, 5)]

print("=== Combined BLEU Evaluation (1–4 merged) ===")

bleu1 = BLEUScore(n_gram=1)
bleu2 = BLEUScore(n_gram=2)

def sanitize(text):
    text = str(text).strip()
    return text if text else "null"

# --------------------------------------------
# Merge all preds/targets from 1–4 into one
# --------------------------------------------
overall_preds = []
overall_tgts = []

for pcol, tcol in zip(pred_cols, tgt_cols):
    preds = [sanitize(row[pcol]) for row in data]
    tgts  = [sanitize(row[tcol]) for row in data]

    overall_preds.extend(preds)
    overall_tgts.extend(tgts)

print(f"Total merged examples: {len(overall_preds)}")

# --------------------------------------------
# Compute ONE combined BLEU score
# --------------------------------------------
try:
    b1 = bleu1(overall_preds, overall_tgts).item()
    b2 = bleu2(overall_preds, overall_tgts).item()

    print("\n=== FINAL COMBINED BLEU SCORE (1–4 merged) ===")
    print("BLEU1:", b1)
    print("BLEU2:", b2)

except Exception as e:
    print("BLEU failed:", e)
```
